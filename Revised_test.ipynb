```python
# 1. Checks student's marks and gives Pass/Fail based on marks

marks = [45,32,78,39,90,28,55,40,67,12]

pass_count = 0   # used for count pass
fail_count = 0   #used for count fail

for mark in marks:
 if mark>=40:
  print("Pass")
  pass_count = pass_count + 1

 else:
   print("Fail")
   fail_count = fail_count + 1

#To display total marks
print("Total pass",pass_count)
print("Total fail",fail_count)
```

```python
# Q2. Safe Average Calculator

# Function to calculate the average of numbers
def calculate_average(numbers: list) -> float:
    """Returns the average of a list of numbers."""
    return sum(numbers) / len(numbers)


try:                                     # Test the function with a normal list
    print("Average:", calculate_average([10, 20, 30]))

except ZeroDivisionError:                # Handle the error if the list is empty
    print("Cannot calculate average of an empty list.")


try:
    print("Average:", calculate_average(numbers=[]))

except ZeroDivisionError:
    print("Cannot calculate average of an empty list.")
```

```python
# 3. Daily Fitness Tracker

import random

steps = []


for i in range(7):                    #step counts for 7 days
    step_count = random.randint(2000, 12000)
    steps.append(step_count)
    print("Day", i + 1, step_count, "steps")

print("Highest steps:", max(steps))    #Find highest and lowest steps
print("Lowest steps:", min(steps))

average = sum(steps) / 7
print("Average steps:", average)
```

```python
# Q4.Calculate avg marks and highest mark student

class Student:
    def __init__(self, name: str, marks: list):
        self.name = name
        self.marks = marks

    def average(self):

        return sum(self.marks) / len(self.marks)

    def __str__(self):
           return self.name


students = [
    Student("Shreyasi", [80, 85, 90]),
    Student("Meera", [70, 75, 80]),
    Student("Priya", [90, 92, 88])
]

for student in students:
    print(student)

highest = students[0]

for student in students:
    if student.average() > highest.average():
        highest = student
print("Highest average:", highest.name)
```
